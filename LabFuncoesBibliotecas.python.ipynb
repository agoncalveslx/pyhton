# 1. Introdução às funções



---

<div style="background-color:#f1f5f9; padding:15px; border-left:6px solid #2563eb; border-radius:6px;">

### Ideia principal

Uma função é um bloco de código reutilizável que ajuda a organizar o programa, evitar repetição e executar uma tarefa sempre que necessário.

</div>

---

## Para que servem?

| Utilidade | Explicação |
|---|---|
| **Evitar repetição** | Reutilizam o mesmo código várias vezes |
| **Organizar o programa** | Dividem o código em partes mais simples |
| **Melhorar a leitura** | Tornam o código mais claro |
| **Facilitar alterações** | Permitem alterar uma tarefa num único local |

---

## Definir e chamar uma função

Uma função tem três partes importantes:

| Parte | Significado |
|---|---|
| **Definição** | Criação da função com `def` |
| **Corpo da função** | Instruções indentadas que pertencem à função |
| **Chamada** | Execução da função pelo seu nome |

```python
def mostrar_mensagem():
    print("Bem-vindo à aula de Python!")

mostrar_mensagem()
```

Neste exemplo, `def mostrar_mensagem():` define a função e `mostrar_mensagem()` chama a função.

---

## A ordem importa?

Sim. Em Python, o código é executado de cima para baixo.  
Por isso, a função deve estar definida antes de ser chamada durante a execução.

| Situação | Resultado |
|---|---|
| Definir antes de chamar | Correto |
| Chamar antes de definir | Erro |

---

## Nota

Em Python, as funções são criadas com a palavra-chave `def`.

As células seguintes mostram um exemplo sem função e depois o mesmo exemplo usando uma função.


````python
# Exemplo simples sem função

print("Bem-vindo à aula de Python!")
print("Bem-vindo à aula de Python!")
print("Bem-vindo à aula de Python!")
````

````python
# O mesmo exemplo usando uma função

def mostrar_mensagem():
    print("Bem-vindo à aula de Python!")

mostrar_mensagem()
mostrar_mensagem()
mostrar_mensagem()
````

# 2. Objetos mutáveis e imutáveis em Python

---

<div style="background-color:#f1f5f9; padding:15px; border-left:6px solid #2563eb; border-radius:6px;">

### Ideia principal

Em Python, os valores são objetos. Algumas variáveis podem referenciar o mesmo objeto, mas o comportamento depende de esse objeto ser mutável ou imutável.

</div>

---

## Tipos de objetos

| Tipo | Explicação | Exemplos |
|---|---|---|
| **Imutável** | Não pode ser alterado diretamente depois de criado | `int`, `float`, `str`, `bool`, `tuple` |
| **Mutável** | Pode ser alterado depois de criado | `list`, `dict`, `set` |

---

## Porque é importante?

Esta distinção é relevante quando os objetos são atribuídos a variáveis ou passados para funções.

| Tipo de objeto | Comportamento geral |
|---|---|
| **Imutável** | A alteração faz a variável referenciar outro objeto |
| **Mutável** | O conteúdo do próprio objeto pode ser alterado |

---

## Demonstração

As células seguintes mostram:

1. um exemplo com um objeto imutável;
2. um exemplo com um objeto mutável.


````python
# Exemplo de objeto imutável
# Os inteiros são imutáveis: o valor 10 não é alterado diretamente.

print("=== Objeto imutável ===")
print()

numero = 10
mesmo_numero = numero

print("Valor inicial através de numero:", numero)
print("Valor inicial através de mesmo_numero:", mesmo_numero)

print()

# numero passa a referenciar um novo objeto com o valor 11.
# mesmo_numero continua a referenciar o valor 10.

numero = numero + 1

print("Depois de fazer numero = numero + 1:")
print("numero:", numero)
print("mesmo_numero:", mesmo_numero)

print()

# mesmo_numero passa a referenciar um novo objeto com o valor 15.
# numero continua a referenciar o valor 11.

mesmo_numero = mesmo_numero + 5

print("Depois de fazer mesmo_numero = mesmo_numero + 5:")
print("numero:", numero)
print("mesmo_numero:", mesmo_numero)

print()

# Conclusão:
# Como os inteiros são imutáveis, cada alteração cria/obtém um novo objeto.
# Cada variável pode passar a referenciar um objeto diferente.
````

````python
# Exemplo de objeto mutável
# As listas são objetos mutáveis: podem ser modificadas depois de criadas.

print("=== Objeto mutável ===")
print()

nomes = ["Ana", "João"]
mesma_lista = nomes

print("Lista inicial através de nomes:", nomes)
print("Lista inicial através de mesma_lista:", mesma_lista)

print()

mesma_lista.append("Maria")

print("Depois de adicionar 'Maria' com mesma_lista:")
print("nomes:", nomes)
print("mesma_lista:", mesma_lista)

print()

nomes.append("Pedro")

print("Depois de adicionar 'Pedro' com nomes:")
print("nomes:", nomes)
print("mesma_lista:", mesma_lista)

print()

# Conclusão:
# nomes e mesma_lista são duas variáveis diferentes,
# mas referenciam o mesmo objeto em memória.
# Como a lista é mutável, a alteração feita através de uma variável
# aparece também através da outra.
````

# 6. `print()` vs `return`

---

<div style="background-color:#f1f5f9; padding:15px; border-left:6px solid #2563eb; border-radius:6px;">

### Ideia principal

`print()` mostra informação no ecrã. `return` devolve um valor ao programa.

</div>

---

## Diferença principal

| Comando | O que faz | Utilização típica |
|---|---|---|
| `print()` | Mostra informação no ecrã | Apresentar resultados ao utilizador |
| `return` | Devolve informação ao programa | Guardar, reutilizar ou processar resultados |

---

## Exemplo conceptual

O factorial de `n` é calculado multiplicando os números de `1` até `n`.

```text
5! = 5 × 4 × 3 × 2 × 1 = 120
```

```python
def mostrar_factorial(n):
    # calcular o factorial
    resultado = ...
    print(resultado)

def calcular_factorial(n):
    # calcular o factorial
    resultado = ...
    return resultado
```

| Exemplo | Objetivo |
|---|---|
| `mostrar_factorial(n)` | Calcula e mostra o resultado com `print()` |
| `calcular_factorial(n)` | Calcula e devolve o resultado com `return` |

---

## Nota importante

Uma função sem `return` devolve automaticamente `None`.  
Quando se pretende reutilizar um resultado, deve ser usado `return`.


# 7. Exemplo utilizado: factorial de um número

---

<div style="background-color:#f1f5f9; padding:15px; border-left:6px solid #2563eb; border-radius:6px;">

### Ideia principal

O factorial será usado para demonstrar a diferença entre um procedimento com `print()` e uma função com `return`.

</div>

---

## Factorial

O factorial de um número inteiro não negativo `n` representa-se por `n!`.

```text
n! = n × (n - 1) × (n - 2) × ... × 1
```

Exemplo:

```text
5! = 5 × 4 × 3 × 2 × 1 = 120
```

Nesta aula, o foco está no uso da função, não no detalhe do cálculo matemático.


````python
# Exemplo de procedimento
# Este procedimento calcula o factorial e mostra o resultado no ecrã.
# Como não usa return, não devolve um valor útil ao programa.

def mostrar_factorial(n):
    resultado = 1

    for i in range(1, n + 1):
        resultado = resultado * i

    print("O factorial de", n, "é", resultado)

valor = mostrar_factorial(5)

print("Valor guardado na variável:", valor)
````

````python
# Exemplo de função
# Esta função calcula o factorial e devolve o resultado com return.
# Como usa return, o valor pode ser guardado e reutilizado.

def calcular_factorial(n):
    resultado = 1

    for i in range(1, n + 1):
        resultado = resultado * i

    return resultado

valor = calcular_factorial(5)

print("Valor guardado na variável:", valor)
````

<div style="background-color:#f1f5f9; padding:15px; border-left:6px solid #2563eb; border-radius:6px;">

# 10. Validação de parâmetros e tratamento de erros

---

<div style="background-color:#f1f5f9; padding:15px; border-left:6px solid #2563eb; border-radius:6px;">

### Ideia principal

Uma função deve verificar se recebeu valores válidos antes de executar o cálculo. Quando recebe valores inválidos, pode lançar uma exceção para sinalizar o erro.

</div>

---

## Porque validar parâmetros?

A validação ajuda a:

| Objetivo | Explicação |
|---|---|
| **Evitar erros** | Impede que a função execute cálculos com dados inadequados |
| **Tornar o código mais seguro** | Reduz comportamentos inesperados |
| **Melhorar a clareza** | Indica claramente quais os valores aceites pela função |

---

## Regras para o factorial

No caso do factorial, vamos aplicar duas regras:

| Regra | Exemplo válido | Exemplo inválido | Exceção adequada |
|---|---|---|---|
| O valor deve ser inteiro | `5` | `2.5`, `"João"` | `TypeError` |
| O valor deve ser maior ou igual a zero | `0`, `5` | `-3` | `ValueError` |

---

## Momentos importantes

| Momento | O que acontece |
|---|---|
| **1. Validação dos dados** | A função usa `if` para verificar se os dados recebidos são válidos |
| **2. Tratamento simples do erro** | A função pode devolver uma mensagem de erro com `return` |
| **3. Lançamento de exceção** | A função pode sinalizar o erro com `raise`, usando exceções como `TypeError` ou `ValueError` |
| **4. Chamada sem tratamento** | Se a função lançar uma exceção e não houver `try` / `except`, o erro aparece diretamente no Colab |
| **5. Chamada com tratamento** | O erro é capturado com `try` / `except` e apresentado de forma controlada |

---

## Exemplo conceptual

Existem várias formas de lidar com valores inválidos dentro de uma função.

---

### 1. Validação simples

A função verifica o erro e devolve uma mensagem.

```python
def calcular_factorial(n):

    if n não for inteiro:
        return "Erro: o valor deve ser um número inteiro."

    if n for menor que zero:
        return "Erro: o número deve ser maior ou igual a zero."

    calcular factorial

    return resultado
```

Nesta abordagem, a função não interrompe o programa, mas pode devolver tipos diferentes de resultado: por exemplo, um número quando corre bem e uma mensagem quando corre mal.

---

### 2. Validação com exceções

A função verifica o erro e lança uma exceção.

```python
def calcular_factorial(n):

    if n não for inteiro:
        raise TypeError("O valor deve ser um número inteiro.")

    if n for menor que zero:
        raise ValueError("O número deve ser maior ou igual a zero.")

    calcular factorial

    return resultado
```

Nesta abordagem, a função devolve o resultado apenas quando os dados são válidos. Quando existe erro, sinaliza o problema com uma exceção.

---

## Comparação

| Abordagem | Como sinaliza o erro | Vantagem | Limitação |
|---|---|---|---|
| **Validação simples** | `return "Erro..."` | Fácil de perceber | Mistura resultados com mensagens de erro |
| **Validação com exceções** | `raise TypeError` ou `raise ValueError` | Mais correta e profissional | Exige compreender `try` / `except` |

---

## Ideia-chave

O `if` serve para verificar a condição.  
Depois, a função pode lidar com o erro de duas formas:

- devolver uma mensagem simples com `return`;
- lançar uma exceção com `raise`.


````python
## 1. Validação simples com return
### A função deteta o erro e devolve uma mensagem.
### Esta abordagem é simples, mas mistura números com mensagens de erro.

def calcular_factorial(n):
    if type(n) != int:
        return "Erro: o valor deve ser um número inteiro."

    if n < 0:
        return "Erro: o número deve ser maior ou igual a zero."

    resultado = 1

    for i in range(1, n + 1):
        resultado = resultado * i

    return resultado


valor = 5
resultado = calcular_factorial(valor)
print("Resultado para", valor, ":\t", resultado)

valor = -3
resultado = calcular_factorial(valor)
print("Resultado para", valor, ":\t", resultado)

valor = 2.5
resultado = calcular_factorial(valor)
print("Resultado para", valor, ":\t", resultado)
````

````python
## 2. Validação com raise
### A função deteta o erro e lança uma exceção.
### Esta abordagem é mais profissional.

def calcular_factorial(n):
    if type(n) != int:
        raise TypeError("O valor deve ser um número inteiro.")

    if n < 0:
        raise ValueError("O número deve ser maior ou igual a zero.")

    resultado = 1

    for i in range(1, n + 1):
        resultado = resultado * i

    return resultado


valor = 5
resultado = calcular_factorial(valor)
print("Factorial de", valor, ":\t", resultado)

valor = -3
resultado = calcular_factorial(valor)
print("Factorial de", valor, ":\t", resultado)

valor = 2.5
resultado = calcular_factorial(valor)
print("Factorial de", valor, ":\t", resultado)
````

````python
# 3. Chamada com tratamento da exceção
# O try tenta executar a função.
# O except trata o erro caso ele ocorra.

def calcular_factorial(n):
    if type(n) != int:
        raise TypeError("O valor deve ser um número inteiro.")

    if n < 0:
        raise ValueError("O número deve ser maior ou igual a zero.")

    resultado = 1

    for i in range(1, n + 1):
        resultado = resultado * i

    return resultado


valores_teste = [5, -3, 2.5]

for valor in valores_teste:
    try:
        resultado = calcular_factorial(valor)
        print("Factorial de", valor, ":\t", resultado)

    except ValueError as erro:
        print("Erro de valor para", valor, ":\t", erro)

    except TypeError as erro:
        print("Erro de tipo para", valor, ":\t", erro)
````

# 11. Impacto dos objetos mutáveis e imutáveis nas funções

---

<div style="background-color:#f1f5f9; padding:15px; border-left:6px solid #2563eb; border-radius:6px;">

### Ideia principal

Quando um objeto é passado para uma função, o comportamento pode ser diferente consoante esse objeto seja mutável ou imutável.

</div>

---

## Comportamento geral

| Tipo de objeto | Exemplos | Comportamento dentro da função |
|---|---|---|
| **Imutável** | `int`, `float`, `str`, `bool`, `tuple` | A função não altera diretamente o objeto original |
| **Mutável** | `list`, `dict`, `set` | A função pode alterar o conteúdo do objeto original |

---

## Porque é importante?

Esta distinção ajuda a evitar alterações inesperadas.

| Situação | Resultado possível |
|---|---|
| Passar um número para uma função | O valor original fora da função não é alterado diretamente |
| Passar uma lista para uma função | A lista original pode ser modificada dentro da função |

---

## Nota

Ao trabalhar com objetos mutáveis, como listas ou dicionários, é importante ter cuidado com alterações feitas dentro das funções.

As células seguintes mostram esta diferença com exemplos simples.


````python
# Exemplo com objeto imutável
# Os inteiros são imutáveis.
# A variável x recebe a referência ao mesmo valor que numero.
# Quando fazemos x = x + 1, x passa a referenciar um novo objeto.
# A variável numero, fora da função, continua igual.

def alterar_numero(x):
    print("Dentro da função, antes da alteração:", x)

    x = x + 1

    print("Dentro da função, depois da alteração:", x)

numero = 10

print("Fora da função, antes da chamada:", numero)

alterar_numero(numero)

print("Fora da função, depois da chamada:", numero)
````

````python
# Exemplo com objeto mutável
# As listas são mutáveis.
# A variável lista recebe a referência à mesma lista que nomes.
# Quando fazemos lista.append("Maria"), a lista original é alterada.
# Por isso, a alteração também é visível fora da função.

def adicionar_nome(lista):
    print("Dentro da função, antes da alteração:", lista)

    lista.append("Maria")

    print("Dentro da função, depois da alteração:", lista)

nomes = ["Ana", "João"]

print("Fora da função, antes da chamada:", nomes)

adicionar_nome(nomes)

print("Fora da função, depois da chamada:", nomes)
````

# 12. Boas práticas na criação de funções

---

<div style="background-color:#f1f5f9; padding:15px; border-left:6px solid #2563eb; border-radius:6px;">

### Ideia principal

Uma boa função deve ter um objetivo claro, um nome compreensível e uma única responsabilidade principal.

</div>

---

## Recomendações principais

| Boa prática | Explicação |
|---|---|
| **Usar nomes claros** | O nome da função deve indicar a tarefa que a função realiza |
| **Aplicar o Princípio da Responsabilidade Única** | Cada função deve ter uma responsabilidade principal e resolver uma tarefa bem definida |
| **Manter a função simples** | A função deve ser fácil de ler, testar e corrigir |
| **Evitar repetição** | Código repetido pode ser colocado numa função reutilizável |
| **Validar parâmetros** | A função deve verificar os dados recebidos quando existe risco de valores inválidos |
| **Evitar efeitos inesperados** | Objetos mutáveis, como listas, devem ser alterados com cuidado |
| **Usar `return` quando necessário** | Se o resultado tiver de ser reutilizado, deve ser devolvido com `return` |

---

## Princípio da Responsabilidade Única

Uma função deve fazer **uma tarefa principal** e fazê-la bem.

Exemplo adequado:

```python
def calcular_factorial(n):
    return resultado
```

Exemplo menos adequado:

```python
def calcular_factorial_mostrar_e_guardar(n):
    # calcula o factorial
    # mostra o resultado
    # guarda o resultado noutro local
```

Neste segundo caso, a função acumula várias responsabilidades.

---

## Exemplo de nomes

| Nome | Avaliação | Motivo |
|---|---|---|
| `calcular_factorial()` | Adequado | Indica claramente o que a função faz |
| `cf()` | Pouco claro | É demasiado abreviado e pouco informativo |

---

## Regra prática

Ao ler o nome da função, deve ser possível perceber a sua finalidade principal.


````python
# Exemplo de função com nome claro

def calcular_quadrado(numero):
    return numero * numero

resultado = calcular_quadrado(4)

print("Quadrado:", resultado)
````

# 14. Aplicar `lambda` a listas com `map()` e `filter()`

---

<div style="background-color:#f1f5f9; padding:15px; border-left:6px solid #2563eb; border-radius:6px;">

### Ideia principal

Uma função `lambda` pode ser usada para definir uma operação simples. Com `map()` e `filter()`, essa operação pode ser aplicada aos elementos de uma lista.

</div>

---

## Dois usos comuns

| Função | Para que serve | Exemplo |
|---|---|---|
| `map()` | Transforma todos os elementos de uma lista | Multiplicar todos os números por `2` |
| `filter()` | Seleciona apenas os elementos que cumprem uma condição | Selecionar apenas números pares |

---

## Exemplo com `map()`

```python
numeros = [1, 2, 3, 4, 5]

dobros = list(map(lambda x: x * 2, numeros))
```

Neste exemplo:

| Parte | Significado |
|---|---|
| `lambda x: x * 2` | Define a transformação: multiplicar cada valor por `2` |
| `map(...)` | Aplica essa transformação a todos os elementos |
| `list(...)` | Converte o resultado numa lista visível |

Resultado esperado:

```python
[2, 4, 6, 8, 10]
```

---

## Exemplo com `filter()`

```python
numeros = [1, 2, 3, 4, 5, 6]

pares = list(filter(lambda x: x % 2 == 0, numeros))
```

Neste exemplo:

| Parte | Significado |
|---|---|
| `lambda x: x % 2 == 0` | Define a condição: verificar se o número é par |
| `filter(...)` | Mantém apenas os elementos que cumprem a condição |
| `list(...)` | Converte o resultado numa lista visível |

Resultado esperado:

```python
[2, 4, 6]
```

---

## Nota sobre programação funcional

O uso de `lambda` com funções como `map()` e `filter()` está associado à **programação funcional**.

Neste estilo, uma função pode ser passada como argumento para outra função.

| Conceito | Exemplo |
|---|---|
| **Função anónima** | `lambda x: x * 2` |
| **Função de ordem superior** | `map()` e `filter()` recebem uma função como argumento |
| **Programação funcional** | Uso de funções para transformar ou filtrar dados |

---

## Onde este estilo é útil?

Este tipo de programação é útil quando se pretende aplicar a mesma regra a vários dados.

| Área | Utilidade |
|---|---|
| **Tratamento de dados** | Transformar, limpar ou filtrar listas de valores |
| **Análise de dados** | Selecionar dados que cumprem determinada condição |
| **Automatização** | Aplicar rapidamente uma regra a vários elementos |
| **Big Data** | Aplicar operações semelhantes a grandes volumes de dados, usando ferramentas próprias |

Em contextos de **Big Data**, a ideia de transformar e filtrar dados é muito comum em ferramentas como Apache Spark.  
Nesta aula, os exemplos são simples, mas a lógica é semelhante: definir uma regra e aplicá-la a vários elementos.

---

## Nota final

Nesta aula, estes conceitos são usados apenas de forma introdutória.


````python
# Exemplo com map()
# Multiplicar todos os valores da lista por 2

numeros = [1, 2, 3, 4, 5]

dobros = list(map(lambda x: x * 2, numeros))

print("Lista original:", numeros)
print("Lista com valores a dobrar:", dobros)
````

````python
# Exemplo com filter()
# Selecionar apenas os números pares da lista

numeros = [1, 2, 3, 4, 5, 6]

pares = list(filter(lambda x: x % 2 == 0, numeros))

print("Lista original:", numeros)
print("Números pares:", pares)
````

# 15. Diferença entre `def` e `lambda`

---

<div style="background-color:#f1f5f9; padding:15px; border-left:6px solid #2563eb; border-radius:6px;">

### Ideia principal

Tanto `def` como `lambda` permitem criar funções, mas são usados em contextos diferentes.

</div>

---

## Comparação principal

| Aspeto | `def` | `lambda` |
|---|---|---|
| **Uso principal** | Funções normais e reutilizáveis | Funções curtas e pontuais |
| **Tamanho** | Pode ter várias instruções | Tem apenas uma expressão |
| **Nome** | Normalmente tem um nome definido | Pode ser anónima |
| **Legibilidade** | Melhor para funções maiores | Melhor para operações simples |
| **Exemplo típico** | Calcular factorial com várias instruções | Multiplicar um valor por `2` |

---

## Exemplo conceptual

```python
def dobrar(x):
    return x * 2

lambda x: x * 2
```

Ambas podem representar a mesma operação simples: devolver o dobro de um valor.

---

## Regra prática

| Situação | Recomendação |
|---|---|
| A função tem várias instruções | Usar `def` |
| A função precisa de validação | Usar `def` |
| A função será reutilizada várias vezes | Usar `def` |
| A operação é curta e simples | Pode ser usada `lambda` |
| A função é usada temporariamente com `map()` ou `filter()` | Pode ser usada `lambda` |

---

## Nota

Para código mais claro e fácil de manter, deve ser usada `lambda` apenas quando a expressão é simples e facilmente compreensível.


````python
# Versão com def

def dobrar(x):
    return x * 2

numeros = [1, 2, 3, 4, 5]

dobros = list(map(dobrar, numeros))

print(dobros)
````

````python
# Versão com lambda

numeros = [1, 2, 3, 4, 5]

dobros = list(map(lambda x: x * 2, numeros))

print(dobros)
````

# Programa completo num único ficheiro: gestão de notas


````python
# Programa completo num único ficheiro
# Exemplo: gestão simples de notas de alunos


# -----------------------------
# Dados do programa
# -----------------------------

alunos = {
    "Ana": [14, 16, 15],
    "João": [9, 11, 10],
    "Maria": [18, 17, 19],
    "Pedro": [12, 25, 13]
}


# -----------------------------
# Modelo: validações e cálculos
# -----------------------------

def validar_nota(nota):
    if type(nota) != int and type(nota) != float:
        raise TypeError("A nota deve ser um número.")

    if nota < 0 or nota > 20:
        raise ValueError("A nota deve estar entre 0 e 20.")


def validar_lista_notas(notas):
    if type(notas) != list:
        raise TypeError("As notas devem estar numa lista.")

    if len(notas) == 0:
        raise ValueError("A lista de notas não pode estar vazia.")

    for nota in notas:
        validar_nota(nota)


def calcular_media(notas):
    validar_lista_notas(notas)

    soma = 0

    for nota in notas:
        soma = soma + nota

    return soma / len(notas)


def obter_nota_maxima(notas):
    validar_lista_notas(notas)

    maior = notas[0]

    for nota in notas:
        if nota > maior:
            maior = nota

    return maior


def obter_nota_minima(notas):
    validar_lista_notas(notas)

    menor = notas[0]

    for nota in notas:
        if nota < menor:
            menor = nota

    return menor


def classificar_media(media):
    if media >= 16:
        return "Muito bom"
    elif media >= 10:
        return "Aprovado"
    else:
        return "Reprovado"


# -----------------------------
# Vista: apresentação dos dados
# -----------------------------

def mostrar_resultado(nome, media, maxima, minima, classificacao):
    print("Aluno:", nome)
    print("Média:", media)
    print("Nota máxima:", maxima)
    print("Nota mínima:", minima)
    print("Classificação:", classificacao)
    print()


def mostrar_erro(nome, erro):
    print("Erro nos dados do aluno:", nome)
    print("Detalhe:", erro)
    print()


def mostrar_alunos_incluidos_e_excluidos(alunos_incluidos, alunos_excluidos):
    print("Alunos incluídos no cálculo da média final:")

    for aluno in alunos_incluidos:
        print("-", aluno)

    print()

    print("Alunos excluídos do cálculo da média final:")

    if len(alunos_excluidos) == 0:
        print("- Nenhum aluno excluído")
    else:
        for aluno in alunos_excluidos:
            print("-", aluno)

    print()


def mostrar_media_turma(media_turma):
    print("Média final da turma com alunos válidos:", media_turma)


# -----------------------------
# Controlador: execução principal
# -----------------------------

medias_validas = []
alunos_incluidos = []
alunos_excluidos = []

for nome, notas in alunos.items():
    try:
        media = calcular_media(notas)
        maxima = obter_nota_maxima(notas)
        minima = obter_nota_minima(notas)
        classificacao = classificar_media(media)

        medias_validas.append(media)
        alunos_incluidos.append(nome)

        mostrar_resultado(nome, media, maxima, minima, classificacao)

    except TypeError as erro:
        alunos_excluidos.append(nome)
        mostrar_erro(nome, erro)

    except ValueError as erro:
        alunos_excluidos.append(nome)
        mostrar_erro(nome, erro)


# -----------------------------
# Síntese final
# -----------------------------

mostrar_alunos_incluidos_e_excluidos(alunos_incluidos, alunos_excluidos)

if len(medias_validas) > 0:
    media_turma = calcular_media(medias_validas)
    mostrar_media_turma(media_turma)
else:
    print("Não existem alunos válidos para calcular a média final da turma.")
````
